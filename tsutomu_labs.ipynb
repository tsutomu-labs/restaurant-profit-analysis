```python
csv_text = """商品,売上,原価率
唐揚げ定食,500000,0.38
生ビール,400000,0.18
ラーメン,350000,0.42
"""

with open("sample_data.csv", "w", encoding="utf-8") as f:
    f.write(csv_text)
```

```python
import pandas as pd

df = pd.read_csv("sample_data.csv")

print(df)
```

```python
df["粗利"] = df["売上"] * (1 - df["原価率"])

print(df)
```

```python
import matplotlib.pyplot as plt

plt.bar(df["商品"], df["粗利"])

plt.title("Gross Profit")

plt.show()
```

```python
df["item_en"] = ["Karaage", "Beer", "Ramen"]

print(df)
```

```python
import matplotlib.pyplot as plt

plt.bar(df["item_en"], df["粗利"])

plt.title("Gross Profit")

plt.show()
```

```python
# 売上順に並び替え
df_abc = df.sort_values(by="売上", ascending=False)

# 売上構成比
df_abc["売上構成比"] = df_abc["売上"] / df_abc["売上"].sum()

# 累積構成比
df_abc["累積構成比"] = df_abc["売上構成比"].cumsum()

# ABCランク判定
def abc_rank(x):
    if x <= 0.7:
        return "A"
    elif x <= 0.9:
        return "B"
    else:
        return "C"

df_abc["ABCランク"] = df_abc["累積構成比"].apply(abc_rank)

print(df_abc)
```

```python
df["粗利率"] = 1 - df["原価率"]

print(df)
```

```python
low_profit = df.sort_values(by="粗利率")

print(low_profit)
```
